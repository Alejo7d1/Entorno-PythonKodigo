Implementa **cuatro funciones** (bucles + condicionales; aquí no se usan excepciones):

1. `eliminar_nulos_y_vacios(lista)` — retorna una lista **nueva** sin `None` y sin cadenas vacías (incluyendo las que quedan vacías tras quitar espacios, como `"  "`).
2. `quitar_espacios(lista)` — retorna la lista con `.strip()` aplicado a los elementos que sean strings (los no-string pasan intactos).
3. `normalizar_texto(lista, formato="titulo")` — retorna la lista con los strings normalizados; el parámetro `formato` acepta `"titulo"`, `"minusculas"` o `"mayusculas"`.
4. `eliminar_duplicados(lista)` — retorna la lista sin duplicados, **manteniendo el primer valor encontrado** y el orden.

Luego **pruébalas** sobre `datos_empresa["nombres"]` y `datos_empresa["productos"]` mostrando el antes y el después (cantidad de elementos y una muestra).

Esta sección es para crear una lista con los valores limpios, eliminando la "basura" que sería toda aquella información que dificulta o incluso podría romper el programa por contener espacios en blanco, saltos de línea, formateo de mayusculas minusculas.

```python


def eliminar_nulos_y_vacios(lista):
    resultado = []
    for elemento in lista:
        # Saltar None directamente
        if elemento is None:
            continue
        # Si es string, verificar que no quede vacío tras strip
        if isinstance(elemento, str) and elemento.strip() == "":
            continue
        resultado.append(elemento)
    return resultado


def quitar_espacios(lista):
    resultado = []
    for elemento in lista:
        if isinstance(elemento, str):
            resultado.append(elemento.strip())
        else:
            resultado.append(elemento)
    return resultado


def normalizar_texto(lista, formato="titulo"):
    resultado = []
    for elemento in lista:
        if isinstance(elemento, str):
            if formato == "titulo":
                resultado.append(elemento.title())
            elif formato == "minusculas":
                resultado.append(elemento.lower())
            elif formato == "mayusculas":
                resultado.append(elemento.upper())
            else:
                resultado.append(elemento)  # formato desconocido, dejar tal cual
        else:
            resultado.append(elemento)
    return resultado


def eliminar_duplicados(lista):
    vistos = set()
    resultado = []
    for elemento in lista:
        # Usar una tupla para hashear listas anidadas si las hay, pero aquí son strings/None
        if elemento not in vistos:
            vistos.add(elemento)
            resultado.append(elemento)
    return resultado
```

```python

# Prueba con nombres
nombres_crudo = datos_empresa["nombres"]
print("PRUEBA nombres:", len(nombres_crudo), "->", nombres_crudo[:8])

# Encadenar funciones: quitar espacios, normalizar, eliminar nulos/vacíos, duplicados
nombres_limpios = eliminar_duplicados(
    eliminar_nulos_y_vacios(normalizar_texto(quitar_espacios(nombres_crudo), "titulo"))
)
print("DESPUÉS:", len(nombres_limpios), "->", nombres_limpios[:8])

# Prueba con productos
productos_crudo = datos_empresa["productos"]
print("PRUEBA productos:", len(productos_crudo), "->", productos_crudo[:8])

productos_limpios = eliminar_duplicados(
    eliminar_nulos_y_vacios(normalizar_texto(quitar_espacios(productos_crudo), "titulo"))
)
print("DESPUÉS:", len(productos_limpios), "->", productos_limpios[:8])
```

## Sección 2 · Funciones de validación con manejo de excepciones — *2.5 pt*

La estrategia EAFP (*Easier to Ask for Forgiveness than Permission*) consiste en intentar realizar una operación directamente y capturar las excepciones si llega a fallar, en lugar de verificar previamente cada condición. Permite manejar errores de conversión y formato, evitando que el programa colapse.

Implementa **cinco funciones validadoras**. Todas deben usar **`try`-`except`** (nunca dejar escapar un error), retornar el **valor convertido** cuando el dato es recuperable y **`None`** cuando no lo es:

1. `validar_edad(dato)` — convierte a entero y acepta solo el rango **0–120**.
2. `validar_correo(dato)` — limpia espacios, pasa a minúsculas y acepta solo si contiene **una** `@` con texto antes, y un dominio con al menos un punto y texto a ambos lados (rechaza `"usuario@"`, `"luis@email"`, `"isabel@.com"`).
3. `validar_salario(dato)` — convierte a `float` y acepta solo **valores positivos**.
4. `validar_calificacion(dato)` — convierte a entero y acepta solo el rango **0–100**.
5. `validar_codigo_postal(dato)` — limpia espacios y acepta solo **exactamente 4 dígitos numéricos** (retórnalo como string, p. ej. `"1101"`).

Luego **pruébalas** sobre sus listas correspondientes de `datos_empresa`, y **demuestra el manejo de errores**: muestra al menos 3 datos inválidos de cada lista y cómo tu función los neutraliza retornando `None` en lugar de colapsar.


```python

import re

def validar_edad(dato):
    #Valida y convierte un dato a entero en el rango 0–120.
    try:
        edad = int(dato)
        if 0 <= edad <= 120:
            return edad
        return None
    except (ValueError, TypeError):
        return None


def validar_correo(dato):
    #Valida un correo electrónico limpiando espacios y pasando a minúsculas.
    try:
        correo = str(dato).strip().lower()
        # Debe contener exactamente una @
        if correo.count("@") != 1:
            return None
        parte_local, dominio = correo.split("@")
        # Texto antes de la @
        if not parte_local:
            return None
        # Dominio debe tener al menos un punto con texto a ambos lados
        if "." not in dominio:
            return None
        # Verificar que no empiece ni termine con punto y haya texto entre puntos
        partes_dominio = dominio.split(".")
        if any(not parte for parte in partes_dominio):
            return None
        return correo
    except (ValueError, TypeError, AttributeError):
        return None


def validar_salario(dato):
    #Valida y convierte un dato a float positivo.
    try:
        salario = float(dato)
        if salario > 0:
            return salario
        return None
    except (ValueError, TypeError):
        return None


def validar_calificacion(dato):
    #Valida y convierte un dato a entero en el rango 0–100.
    try:
        calif = int(dato)
        if 0 <= calif <= 100:
            return calif
        return None
    except (ValueError, TypeError):
        return None


def validar_codigo_postal(dato):
    #Valida que el dato sea exactamente 4 dígitos numéricos.
    try:
        cp = str(dato).strip()
        if re.fullmatch(r"\d{4}", cp):
            return cp
        return None
    except (ValueError, TypeError, AttributeError):
        return None
```

```python

def probar_validador(lista, validador, nombre):
    #Aplica un validador a una lista y muestra estadísticas + ejemplos de errores
    validados = [validador(x) for x in lista]
    validos = [v for v in validados if v is not None]
    rechazados = [lista[i] for i in range(len(lista)) if validados[i] is None]
    print(f"VALIDADOR: {nombre} | válidos: {len(validos)} | rechazados: {len(rechazados)}")
    print("  3 inválidos neutralizados:", rechazados[:3], "-> None")
    return validados


# Aplicar a cada lista
edades_validadas = probar_validador(datos_empresa["edades"], validar_edad, "validar_edad")
correos_validados = probar_validador(datos_empresa["correos"], validar_correo, "validar_correo")
salarios_validados = probar_validador(datos_empresa["salarios"], validar_salario, "validar_salario")
calificaciones_validadas = probar_validador(datos_empresa["calificaciones"], validar_calificacion, "validar_calificacion")
codigos_validados = probar_validador(datos_empresa["codigos_postales"], validar_codigo_postal, "validar_codigo_postal")
```

## Sección 3 · Funciones de transformación con lambda — *1.5 pt*

Una función lambda es una función de una sola expresión que se define en línea con lambda argumentos: expresión, sin necesidad de def ni return explícito. Se usa cuando la lógica es breve y se necesita pasar una función como argumento

Implementa **al menos tres transformaciones**; **mínimo dos deben usar `lambda`** (con `map()`, `filter()` o `sorted()`):

1. `convertir_a_tipo(lista, tipo)` — recibe una lista de valores mixtos y un tipo destino (`int` o `float`); usa `map()` con una lambda que convierta lo convertible y retorna solo los valores convertidos.
2. `ajustar_salarios(lista_salarios, porcentaje)` — aplica un incremento (o descuento si es negativo) porcentual a una lista de salarios **ya limpios**, usando `map()` + `lambda`. Redondea a 2 decimales.
3. `categorizar_edades(lista_edades)` — usa `map()` + `lambda` + operador ternario para etiquetar edades **ya limpias**: `"joven"` (< 30), `"adulto"` (30–59), `"adulto mayor"` (60+).
4. `filtrar_por_criterio(lista, criterio)` — recibe una función/lambda como parámetro `criterio` y usa `filter()` para retornar los elementos que lo cumplen (pruébala filtrando productos que empiecen con una letra, o departamentos iguales a `"Ventas"`).

**Prueba** cada transformación mostrando entrada y salida.


```python

from functools import reduce


def convertir_a_tipo(lista, tipo):
    #Convierte a un tipo destino los valores convertibles de una lista.

    def _convertir(x):
        try:
            if tipo == "int":
                v = int(float(x)) if isinstance(x, str) else int(x)
                return v if v >= 0 else None
            elif tipo == "float":
                v = float(x)
                return v if v >= 0 else None
            return None
        except (ValueError, TypeError):
            return None

    # map() con lambda que aplica la conversión segura
    resultado = list(filter(lambda v: v is not None, map(lambda x: _convertir(x), lista)))
    return resultado


def ajustar_salarios(lista_salarios, porcentaje):
    # Aplica un cambio porcentual a salarios ya limpios.

    # map() + lambda: nuevo_salario = salario * (1 + porcentaje/100)
    return list(map(lambda s: round(s * (1 + porcentaje / 100), 2), lista_salarios))


def categorizar_edades(lista_edades):
    # Etiqueta edades ya limpias según su categoría de edad.
    # map() + lambda + operador ternario
    return list(map(
        lambda e: "joven" if e < 30 else ("adulto" if e <= 59 else "adulto mayor"),
        lista_edades
    ))


def filtrar_por_criterio(lista, criterio):
    #Retorna los elementos de la lista que cumplen el criterio dado.
    return list(filter(criterio, lista))
```

## Sección 4 · Procesamiento con estructuras de control — *2.0 pt*

En esta sección se convina todo lo anterior. Primero se genera un reporte de calidad que recorre cada conjunto de datos_empresa y cuenta cuántos registros son válidos e inválidos aplicando la función de limpieza correspondiente. Construimos datos_limpios, un nuevo diccionario con la versión limpia de cada lista. Y finalmente calculo **estadísticas básicas** (mínimo, máximo y promedio) sobre edades, salarios y calificaciones ya limpias.

Aquí se orquesta todo lo anterior. Implementa:

1. `generar_reporte_calidad(datos)` — itera sobre el diccionario `datos_empresa` con un bucle y retorna/imprime, **por cada conjunto**: total de registros, cuántos son válidos y cuántos inválidos (usa tus funciones de las secciones 1 y 2 según el conjunto).
2. `limpiar_todo(datos)` — retorna un **nuevo diccionario** `datos_limpios` con la versión limpia de todas las listas (aplica a cada conjunto las funciones que le correspondan: texto → limpieza básica; numéricos y correos → validadores).
3. `estadisticas_basicas(lista_numerica)` — retorna (mínimo, máximo, promedio) de una lista numérica ya limpia. Aplícala al menos a **edades**, **salarios** y **calificaciones** limpias, presentando los resultados con f-strings formateadas.


```python

LIMPIADORES = {
    "nombres":          lambda lista: eliminar_duplicados(eliminar_nulos_y_vacios(normalizar_texto(quitar_espacios(lista), "titulo"))),
    "productos":        lambda lista: eliminar_duplicados(eliminar_nulos_y_vacios(normalizar_texto(quitar_espacios(lista), "titulo"))),
    "departamentos":    lambda lista: eliminar_duplicados(eliminar_nulos_y_vacios(normalizar_texto(quitar_espacios(lista), "titulo"))),
    "edades":           lambda lista: [validar_edad(x) for x in lista],
    "salarios":         lambda lista: [validar_salario(x) for x in lista],
    "calificaciones":   lambda lista: [validar_calificacion(x) for x in lista],
    "codigos_postales": lambda lista: [validar_codigo_postal(x) for x in lista],
    "correos":          lambda lista: [validar_correo(x) for x in lista],
    "cantidades":       lambda lista: convertir_a_tipo(lista, "int"),
    "fechas":           lambda lista: eliminar_nulos_y_vacios(quitar_espacios(lista)),
}


def generar_reporte_calidad(datos):
    # Genera e imprime un reporte de calidad por cada conjunto de datos.
    reporte = {}
    print("=" * 60)
    print("REPORTE DE CALIDAD DE DATOS")
    print("=" * 60)
    print(f"{'Conjunto':<16} {'Total':>6} {'Válidos':>8} {'Inválidos':>9} {'% Rechazo':>10}")
    print("-" * 60)

    for clave, lista in datos.items():
        limpiador = LIMPIADORES[clave]
        # Limpiar y contar cuántos valores quedaron válidos (no None / no vacíos)
        limpios = limpiador(lista)
        validos = len(limpios)
        total = len(lista)
        invalidos = total - validos
        pct = (invalidos / total * 100) if total else 0
        reporte[clave] = (total, validos, invalidos)
        print(f" {clave:<15} {total:>6} {validos:>8} {invalidos:>9} {pct:>9.1f}%")

    print("-" * 60)
    return reporte


def limpiar_todo(datos):
    # Construye un nuevo diccionario con la versión limpia de cada lista.

    datos_limpios = {}
    for clave, lista in datos.items():
        limpiador = LIMPIADORES[clave]
        # Filtrar valores inválidos (None) para dejar solo datos utilizables
        limpios = limpiador(lista)
        datos_limpios[clave] = [v for v in limpios if v is not None]
    return datos_limpios


def estadisticas_basicas(lista_numerica):
    # Calcula mínimo, máximo y promedio de una lista numérica ya limpia.
    if not lista_numerica:
        return (0, 0, 0.0)
    minimo = min(lista_numerica)
    maximo = max(lista_numerica)
    promedio = sum(lista_numerica) / len(lista_numerica)
    return (minimo, maximo, promedio)
```

```python

# Reporte de calidad: válidos vs inválidos por conjunto
reporte = generar_reporte_calidad(datos_empresa)

# Construir el diccionario de datos limpios
datos_limpios = limpiar_todo(datos_empresa)
print("DATOS LIMPIOS:", {k: len(v) for k, v in datos_limpios.items()})

# Estadísticas básicas: edades, salarios y calificaciones limpias
for nombre, lista in [
    ("Edades", datos_limpios["edades"]),
    ("Salarios", datos_limpios["salarios"]),
    ("Calificaciones", datos_limpios["calificaciones"]),
]:
    minimo, maximo, promedio = estadisticas_basicas(lista)
    print(f"{nombre}: min={minimo}, max={maximo}, promedio={promedio:.2f}")
```
