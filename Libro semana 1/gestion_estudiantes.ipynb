# Generación de Listas y diccionarios

Creación de lista de estudiantes


```python
# Lista global que almacenará todos los estudiantes registrados
lista_estudiantes = []
```

Función de agregar estudiante, insertando un diccionario

```python
def agregar_estudiante(nombre, edad, calificaciones):
    # Crea un diccionario con los datos del estudiante
    nuevo_estudiante = {
        "nombre": nombre,
        "edad": edad,
        "calificaciones": calificaciones
    }
    # Añade el estudiante a la lista global
    lista_estudiantes.append(nuevo_estudiante)
    print(f"Estudiantes agregado {nuevo_estudiante}")

# Registro de estudiantes de ejemplo
agregar_estudiante("juan", 25, [7,8,9,4])
agregar_estudiante("Shakira", 75, [4,10,9,2])
agregar_estudiante("Gustanvo", 30, [10,10,9,8])
agregar_estudiante("Paco", 48, [0,2,4,1])
```

Función de buscar el promedio de un estudiante

```python
def calcular_promedio(nombre_estudiante):
    # Busca el estudiante por nombre en la lista
    estudiante = None
    for x in lista_estudiantes:
        if nombre_estudiante == x["nombre"]:
            estudiante = x
            break
    else:
        print("Estudiante no encontrado")
        return 0

    # Calcula el promedio de sus calificaciones
    notas = estudiante["calificaciones"]
    if len(notas) == 0:
        return 0
    return (sum(notas) / len(notas))

# Prueba con el estudiante "juan"
calcular_promedio("juan")
```

Estudiante con mejor promedio

```python
def mejor_promedio():
    mejor = 0
    estudiante_MP = None
    # Recorre la lista buscando el mayor promedio
    for x in lista_estudiantes:
        if calcular_promedio(x["nombre"]) > mejor:
            mejor = calcular_promedio(x["nombre"])
            estudiante_MP = x

    if mejor == 0:
        print("No hay estudiantes registrados o todos tienen 0")

    print(f"El estudiante con mejor promedio es: {estudiante_MP['nombre']}' con {calcular_promedio(estudiante_MP['nombre'])}")

mejor_promedio()
```

Estudiante con peor promedio

```python
def peor_promedio():
    peor = 10
    estudiante_PP = None
    # Recorre la lista buscando el menor promedio
    for x in lista_estudiantes:
        if calcular_promedio(x["nombre"]) < peor:
            peor = calcular_promedio(x["nombre"])
            estudiante_PP = x

    if peor == 11:
        print("No hay estudiantes registrados o todos tienen 10")

    print(f"El estudiante con mejor promedio es: {estudiante_PP['nombre']}' con {calcular_promedio(estudiante_PP['nombre'])}")

peor_promedio()
```

Promedio general del grupo

```python
def promedio_grupo():
    q_estudiantes = len(lista_estudiantes)
    sum_promedios = 0

    # Suma los promedios individuales de todos los estudiantes
    for x in lista_estudiantes:
        sum_promedios = calcular_promedio(x["nombre"]) + sum_promedios

    # Retorna el promedio general del grupo
    return sum_promedios / q_estudiantes

print(f"El promedio del grupo es {promedio_grupo()}")
```
